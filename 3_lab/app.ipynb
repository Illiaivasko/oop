### 1. Оголошення та реалізація класу MyName
Клас містить змінну класу `total_names`, конструктор `__init__` з валідацією та капіталізацією імені, властивості `whoami`, `my_email`, `full_name`, методи `create_email()`, `count_letters()`, `save_to_file()`, класовий метод `anonymous_user()` та статичний метод `say_hello()`.

```python
class MyName:
    """Опис класу / Документація роботи з об'єктами класу MyName"""
    total_names = 0  # Змінна класу (Class Variable)

    def __init__(self, name=None) -> None:
        """Ініціалізація екземпляра класу з валідацією та автоматичною капіталізацією"""
        if name is not None:
            if not isinstance(name, str) or not name.isalpha():
                raise ValueError("Ім'я може містити лише літери!")
            self.name = name.capitalize()
        else:
            self.name = self.anonymous_user().name

        MyName.total_names += 1
        self.my_id = self.total_names

    @property
    def whoami(self) -> str:
        """Властивість класу для отримання представлення імені"""
        return f"My name is {self.name}"

    @property
    def my_email(self) -> str:
        """Властивість класу для отримання поштової скриньки"""
        return self.create_email()

    def create_email(self, domain: str = "itcollege.lviv.ua") -> str:
        """Метод екземпляра з динамічним доменом пошти"""
        return f"{self.name.lower()}@{domain}"

    def count_letters(self) -> int:
        """Метод підрахунку кількості літер в імені"""
        return len(self.name)

    @property
    def full_name(self) -> str:
        """Властивість повного представлення користувача"""
        return f"User #{self.my_id}: {self.name} ({self.my_email})"

    def save_to_file(self, filename: str = "users.txt") -> None:
        """Метод збереження запису користувача у файл"""
        with open(filename, "a", encoding="utf-8") as f:
            f.write(f"{self.full_name}\n")

    @classmethod
    def anonymous_user(cls):
        """Метод класу для створення анонімного користувача"""
        return cls("Anonymous")

    @staticmethod
    def say_hello(message="Hello to everyone!") -> str:
        """Статичний метод з кастомним привітанням"""
        return f"You say: {message}"
```

### 2. Створення об'єктів та демонстрація роботи
Додаємо власне ім'я `Illia` до кортежу `names` та генеруємо об'єкти за допомогою словникового виразу (dict comprehension).

```python
print("Розпочинаємо створювати обєкти!")
names = ("Bohdan", "Marta", "Illia", None)
all_names = {name: MyName(name) for name in names}

for name, me in all_names.items():
    print(f"""{">*<"*20}
This is object: {me} 
This is object attribute: {me.name} / {me.my_id}
This is {type(MyName.whoami)}: {me.whoami} / {me.my_email}
This is {type(me.create_email)} call: {me.create_email()}
This is static {type(MyName.say_hello)} with defaults: {me.say_hello()} 
This is class variable {type(MyName.total_names)}: from class {MyName.total_names} / from object {me.total_names}
{"<*>"*20}""")
```

### 3. Демонстрація виконання індивідуальних завдань

```python
# 1. Кастомне привітання
user = all_names["Illia"]
print("1. Кастомне привітання:")
print("  ", user.say_hello("Доброго дня викладачу Бугель Ю.О.!"))

# 2. Підрахунок літер
print("\n2. Кількість літер:")
for n, obj in all_names.items():
    print(f"   {obj.name} -> {obj.count_letters()}")

# 3. Автоматична капіталізація
user_lower = MyName("taras")
print(f"\n3. Автоматична капіталізація:\n   Введено 'taras' -> Отримано: {user_lower.name}")

# 4. Динамічний email
print(f"\n4. Динамічний email домен:\n   {user.create_email()}\n   {user.create_email('lpnu.ua')}")

# 5. full_name та збереження
print(f"\n5. Властивість full_name та збереження:\n   {user.full_name}")
user.save_to_file("users.txt")
print("   Запис додано у файл users.txt")

# 6. Валідація некоректного імені
print("\n6. Валідація некоректного імені:")
try:
    bad_user = MyName("User123")
except ValueError as e:
    print(f"   Успішно перехоплено помилку: {e}")
```
